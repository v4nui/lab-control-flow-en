## Words

```python
words = ['play', 'filling', 'bar', 'theatre', 'easygoing', 'date', 'lead', 'that', 'story',  'island']
```

**Print every word in upper case**

```python
# your code here
for i in words:
  print(i.upper())
```

**Create a new list containing only words with 5 or more letters**

```python
# your code here
list_5 = []

for i in range(len(words)):   #number of iterations is the length of the list
    if len(words[i]) >=  5:
        list_5.append(words[i])

print(list_5)
```

**Print the first word starting with "t"**

```python
# your code here
i = 0

while i < len(words):
    if words[i].startswith("t"):
        print(words[i])
        break
    i += 1
```

## Numbers

**Create a list containing the square of every number from 1 to 10**

```python
# your code here
sqr_ls = []

for i in range(1,11):
    sqr_ls.append(i ** 2)

print(sqr_ls)
```

**Print a list containing the square of every odd number from 1 to 10**

```python
# your code here
sqr_ls_odd =  []

for i in range(1,11):
    if i % 2 != 0:
        sqr_ls_odd.append(i ** 2)

print(sqr_ls_odd)
```

**Create a list with the squares of all multiples of 8 below 1000**

```python
# your code here
sqr_ls_big = []

for i in range(8,1001):
    if i % 8 == 0:
        sqr_ls_big.append(i ** 2)

print(sqr_ls_big)
```

## People

```python
people = [
    {
        "name": "Juan",
        "age": 34,
        "n_kids": 2
    },
    {
        "name": "Pepe",
        "age": 27,
        "n_kids": 0
    },
    {
        "name": "Sonia",
        "age": 41,
        "n_kids": 1
    },
    {
        "name": "Lucía",
        "age": 22,
        "n_kids": 2
    },
    {
        "name": "Leo",
        "age": 55,
        "n_kids": 5
    }
]
```

**How many people are there?**

```python
# your code here
len(people)
```

**How many people have kids**?

```python
# your code here
j = 0

for i in range(len(people)):
    if people[i]["n_kids"] > 0:  #checking i-th elements number of kids
        j += 1

print(j)
```

**How many kids do they have in total?**

```python
# your code here
total_kids = 0 #int for the summary of kids

for i in range(len(people)):
    if people[i]["n_kids"] > 0:
        total_kids += people[i]["n_kids"]

print(total_kids)
```

**In a year's time, names ending with "a" will have an extra kid. Create a list of dictionaries with people's info in a year's time**

```python
# your code here
people_updated = []

for i in range(len(people)):
    if people[i]["name"].endswith("a"):
        people[i]["n_kids"] = people[i]["n_kids"] + 1
        people_updated.append(people[i])

print(people_updated)
```
